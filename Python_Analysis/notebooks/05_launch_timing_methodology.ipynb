## Information boundary

A launch recommendation may use:
- release month/date
- historical genre performance
- historical platform performance
- historical competitive release density
- platform lifecycle position

It must not use:
- future sales
- future reviews
- post-launch critic scores
- future platform performance
- future competitor outcomes

This distinction keeps timing analysis compatible with the pre-launch prediction model.
